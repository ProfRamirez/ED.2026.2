```python
"""
UFRJ - Universidade Federal do Rio de Janeiro
CCMN - Centro de Ciência Matemáticas e da Natureza
IM   - Instituto de Matemática

Disciplina: Estrutura de Dados (MAE014 - ED.2026.2)
Prof.     : Milton R Ramirez (milton@matematica.ufrj.br)

Data Criação  : 05 de novembro de 2026
Data Alteração: 12 de novembro de 2026

Exemplos de implementação de arvore binárias de busca (BST)
"""
"""
A partir da implementação da Arvore Binária Geral,
produzir o código da arvore binária de busca
"""
class Dado:
  def __init__(self, chave, numero_da_linha_no_arquivo):
    self.chave = chave
    self.linha_do_arquivo = numero_da_linha_no_arquivo

class NoArvore:
  def __init__(self, dado):
    self.dado = dado
    self.esq = None
    self.dir = None

class ArvoreBinariaBusca:
  def __init__(self):
    self.raiz = None

  def insere(self, chave, no_da_linha, no = None):
    if self.raiz == None:
      self.raiz = NoArvore(Dado(chave,no_da_linha))
    else:
      if no == None:
        no = self.raiz
      if chave < no.dado.chave:
        if no.esq == None:
          no.esq = NoArvore(Dado(chave,no_da_linha))
        else:
          self.insere(chave, no_da_linha, no.esq)
      else:
        if no.dir == None:
          no.dir = NoArvore(Dado(chave,no_da_linha))
        else:
          self.insere(chave, no_da_linha, no.dir)

  def busca(self, chave, no = None):
    if no == None:
      no = self.raiz
    if no.dado.chave == chave:
      return no.dado
    else:
      if chave < no.dado.chave:
        if no.esq == None:
          return None
        return self.busca(chave, no.esq)
      else:
        if no.dir == None:
          return None
        return self.busca(chave, no.dir)


abb = ArvoreBinariaBusca()
abb.insere(10,1)
abb.insere(5,2)
abb.insere(15,3)
abb.insere(3,4)
abb.insere(7,5)
abb.insere(13,6)
abb.insere(17,7)

print(abb.busca(10).linha_do_arquivo)
print(abb.busca(100).linha_do_arquivo if abb.busca(100) else "nao encontrado")

"""
1. Inserção de um par (chave, numero da linha) na ABB,

2. Remoção de um no identificado por uma “chave”.

3. Percurso em ordem (ou ordem simétrica) para listar os registros do arquivo
   ordenadamente

"""
```
